```python
import pandas as pd
```

```python
file_path = "CounterFactual Sentence Pairs (1).xlsx"

df = pd.read_excel(file_path, sheet_name="Counterfactual Cleaned")

print("Rows and columns:", df.shape)
print(df.columns.tolist())
df.head()

word_pairs_file_path = "irish_word_pairs_working.xlsx"

df_word_pairs_filtered = pd.read_excel(
    word_pairs_file_path,
    sheet_name="2. Filtered_Dataset"
)

df_word_pairs_native = pd.read_excel(
    word_pairs_file_path,
    sheet_name="3.Adding_Irish_Native_Additions"
)

df_word_pairs_filtered = df_word_pairs_filtered[
    ["Original Terms GA", "Inclusive Terms GA"]
].dropna(
    subset=["Original Terms GA", "Inclusive Terms GA"]
)

df_word_pairs_native = df_word_pairs_native[
    ["Original Terms GA", "Inclusive Terms GA"]
].dropna(
    subset=["Original Terms GA", "Inclusive Terms GA"]
)

df_word_pairs = pd.concat(
    [df_word_pairs_filtered, df_word_pairs_native],
    ignore_index=True
)

print("Total Irish word pairs:", len(df_word_pairs))
df_word_pairs.head()
```

```python
# Prepare data for DataFrame
original_sentences_combined = []
counterfactual_sentences_combined = []
```

```python
# Generate sentences from cleaned Irish counterfactual pairs
for _, row in df.iterrows():
    # Create original sentence
    original_sentence = row["Biased Sentence-Gaeilge"]

    # Create counterfactual sentence
    counterfactual_sentence = row["Counterfactual Sentence-Gaeilge"]

    # Append to lists
    original_sentences_combined.append(original_sentence)
    counterfactual_sentences_combined.append(counterfactual_sentence)
```

```python
# Create DataFrame
data = {
    'Original Sentence': original_sentences_combined,
    'Counterfactual Sentence': counterfactual_sentences_combined
}

df_irish = pd.DataFrame(data)

df_irish.head()
```

```python
df_irish.to_excel(
    'irish_counterfactual_sentences_with_gender_identity.xlsx',
    index=False
)

print(
    "Excel file with Irish original and counterfactual sentences "
    "created successfully."
)
```

```python
with open("Irish_RAG_Pairs.txt", "w", encoding="utf-8") as f:

    # Add Irish word pairs
    for _, row in df_word_pairs.iterrows():
        f.write(
            "Original Term: "
            + str(row["Original Terms GA"])
            + "\n"
        )
        f.write(
            "Inclusive Term: "
            + str(row["Inclusive Terms GA"])
            + "\n"
        )
        f.write("\n")

    # Add Irish counterfactual sentence pairs
    for _, row in df_irish.iterrows():
        f.write(
            "Original Sentence: "
            + str(row["Original Sentence"])
            + "\n"
        )
        f.write(
            "Counterfactual Sentence: "
            + str(row["Counterfactual Sentence"])
            + "\n"
        )
        f.write("\n")

print("Irish RAG text file created successfully.")
```
