```python
df = pd.read_excel("Grammar_Rules.xlsx")

print(df.shape)
print(df.columns.tolist())
df.head()
```

```python
triples = []

for _, row in df.iterrows():
    triple = (
        row["Context"],     # subject/head
        row["Relation"],    # predicate/relation
        row["Constraint"]   # object/tail
    )
    triples.append(triple)

for triple in triples:
    print(triple)
```

NetwrokX


```python
import networkx as nx
```

```python
G = nx.DiGraph()

for _, row in df.iterrows():
    context = row["Context"]
    constraint = row["Constraint"]

    G.add_node(
        context,
        node_type="GrammarContext",
        category=row["Grammar Category"]
    )

    G.add_node(
        constraint,
        node_type="GrammarConstraint"
    )

    G.add_edge(
        context,
        constraint,
        predicate=row["Relation"],
        rule_id=row["Rule ID"],
        example=row["Example"],
        llm_rule_text=row["LLM Rule Text"]

    )

print("Number of nodes:", G.number_of_nodes())
print("Number of edges:", G.number_of_edges())
```

```python
#Test Case
for source, target, data in G.edges(data=True):
    if data["rule_id"] == "GR08":
        print("HEAD:", source)
        print("RELATION:", data["predicate"])
        print("TAIL:", target)
        print("EXAMPLE:", data["example"])
        print("LLM RULE:", data["llm_rule_text"])
```
